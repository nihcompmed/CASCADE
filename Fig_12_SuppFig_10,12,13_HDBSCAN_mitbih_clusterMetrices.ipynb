```python
#!/usr/bin/env python
# coding: utf-8

# ==========================================================
# LOW-PERFORMING PATIENT ANALYSIS
# Goal: demonstrate that N and A are NOT separable for
# patients 118,200,201,203,207,208,213,222
# Uses multiple evidence types so the argument is airtight
# ==========================================================

import os
import numpy as np
import wfdb
import hdbscan
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.patches import Patch
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import adjusted_rand_score, silhouette_score
from scipy.spatial.distance import cdist
from scipy.stats import mannwhitneyu, ks_2samp
from numpy.random import default_rng
from collections import Counter

# ==========================================================
# CONFIG
# ==========================================================

data_folder       = 'mitdb_data_full'
annotation_folder = 'mit-bih-arrhythmia-database-1.0.0'

NORMAL_BEATS     = ['.', 'N', 'L', 'R', 'e', 'j']
ARRHYTHMIC_BEATS = ['A', 'a', 'J', 'S', 'V', 'E', 'F']

beat_length     = 180
num_train_beats = 500
num_val_beats   = 100
num_test_norm   = 100
num_test_arr    = 100
data_seed       = 17

# Good vs poor patients from your results
GOOD_PATIENTS = ['119', '116', '221', '228', '106', '215', '209', '210', '233', '223', '214']
POOR_PATIENTS = ['118', '200', '201', '203', '207', '208', '213', '222']
ALL_PATIENTS  = GOOD_PATIENTS + POOR_PATIENTS

BEST_MCS = 50
BEST_MS  = 10
pca_components = 42

CLR_N     = "#2166ac"
CLR_A     = "#d6604d"
CLR_NOISE = "#cccccc"

fig_dir = "figures_low_perf_analysis"
os.makedirs(fig_dir, exist_ok=True)

# ==========================================================
# HELPERS
# ==========================================================

def get_segments(signal, ann_samples, ann_symbols, beat_length, target_symbols):
    segs, idxs = [], []
    for s, sym in zip(ann_samples, ann_symbols):
        if sym in target_symbols and s >= beat_length:
            segs.append(signal[s - beat_length : s])
            idxs.append(s)
    return segs, idxs

def select_test_beats(n_total, n_train, n_val, n_test, seed):
    rng = default_rng(seed)
    idx = np.arange(n_total)
    rng.shuffle(idx)
    train = idx[n_val : n_val + n_train]
    val   = idx[:n_val]
    test  = idx[n_val + n_train : n_val + n_train + n_test]
    return train, val, test

def mahalanobis_overlap(X_N, X_A):
    """
    Class overlap score: 0 = perfectly separated, 1 = fully overlapping.
    Uses centroid distance normalized by pooled std.
    """
    mu_N, mu_A = X_N.mean(axis=0), X_A.mean(axis=0)
    pooled_std  = np.sqrt((X_N.var(axis=0) + X_A.var(axis=0)) / 2 + 1e-8)
    dist = np.linalg.norm((mu_N - mu_A) / pooled_std)
    overlap = 1 / (1 + dist)   # 1=total overlap, 0=far apart
    return float(overlap), float(dist)

def morphology_stats(X_N, X_A):
    """
    Per-timepoint Mann-Whitney U test between N and A.
    Returns fraction of timepoints where distributions differ (p<0.05).
    """
    n_sig = 0
    for t in range(X_N.shape[1]):
        _, p = mannwhitneyu(X_N[:, t], X_A[:, t], alternative='two-sided')
        if p < 0.05:
            n_sig += 1
    return n_sig / X_N.shape[1]

def within_between_ratio(X, labels):
    """
    Ratio of within-class variance to between-class variance.
    High ratio → classes overlap. Low ratio → well separated.
    """
    classes = np.unique(labels)
    overall_mean = X.mean(axis=0)
    within  = sum(np.mean(np.sum((X[labels==c] - X[labels==c].mean(axis=0))**2, axis=1))
                  for c in classes) / len(classes)
    between = sum(len(X[labels==c]) * np.sum((X[labels==c].mean(axis=0) - overall_mean)**2)
                  for c in classes) / len(X)
    return within / (between + 1e-8)

# ==========================================================
# LOAD AND PROCESS ALL PATIENTS
# ==========================================================

patient_data = {}

for patient in ALL_PATIENTS:
    print(f"Loading {patient}...")
    try:
        ann    = wfdb.rdann(os.path.join(annotation_folder, patient), extension='atr')
        signal = np.load(os.path.join(data_folder, f'{patient}.npy'))[:, 0]
    except Exception as e:
        print(f"  Skipping: {e}")
        continue

    norm_segs, _ = get_segments(signal, ann.sample, ann.symbol, beat_length, NORMAL_BEATS)
    arr_segs,  _ = get_segments(signal, ann.sample, ann.symbol, beat_length, ARRHYTHMIC_BEATS)

    arr_types = sorted(set(sym for sym in ann.symbol if sym in ARRHYTHMIC_BEATS))

    _, _, test_norm_idx = select_test_beats(
        len(norm_segs), num_train_beats, num_val_beats, num_test_norm, seed=data_seed
    )

    test_norm_segs = [norm_segs[i] for i in test_norm_idx]
    test_arr_segs  = arr_segs[:num_test_arr]

    X_raw    = np.array(test_norm_segs + test_arr_segs)
    y_labels = np.array(['N'] * num_test_norm + ['A'] * num_test_arr)
    X_N_raw  = X_raw[:num_test_norm]
    X_A_raw  = X_raw[num_test_norm:]

    # Preprocess
    X_scaled = StandardScaler().fit_transform(X_raw)
    n_comp   = min(pca_components, X_scaled.shape[0]-1, X_scaled.shape[1])
    pca      = PCA(n_components=n_comp)
    X_pca    = pca.fit_transform(X_scaled)

    # HDBSCAN
    clust = hdbscan.HDBSCAN(
        min_cluster_size=BEST_MCS, min_samples=BEST_MS,
        metric='euclidean', cluster_selection_method='leaf'
    )
    clusters = clust.fit_predict(X_pca)

    # t-SNE
    tsne = TSNE(n_components=2, random_state=42, perplexity=20, n_iter=1000)
    X_2d = tsne.fit_transform(X_pca)

    # Metrics
    ari       = adjusted_rand_score(y_labels, clusters)
    noise_frac = np.mean(clusters == -1)

    # Silhouette on PCA (ground truth labels, not cluster labels)
    sil = silhouette_score(X_pca, y_labels, metric='euclidean')

    # Overlap metrics
    X_N_pca = X_pca[:num_test_norm]
    X_A_pca = X_pca[num_test_norm:]
    overlap, centroid_dist = mahalanobis_overlap(X_N_pca, X_A_pca)

    # Morphology: fraction of timepoints significantly different
    frac_sig = morphology_stats(X_N_raw, X_A_raw)

    # Within/between ratio in PCA space
    wb_ratio = within_between_ratio(X_pca, y_labels)

    # Mean waveform per class (raw)
    mean_N = X_N_raw.mean(axis=0)
    mean_A = X_A_raw.mean(axis=0)
    std_N  = X_N_raw.std(axis=0)
    std_A  = X_A_raw.std(axis=0)

    patient_data[patient] = {
        "group":         "Good" if patient in GOOD_PATIENTS else "Poor",
        "arr_types":     ','.join(arr_types),
        "ARI":           ari,
        "silhouette":    sil,
        "overlap":       overlap,
        "centroid_dist": centroid_dist,
        "frac_sig_tp":   frac_sig,
        "wb_ratio":      wb_ratio,
        "noise_frac":    noise_frac,
        "X_2d":          X_2d,
        "X_pca":         X_pca,
        "clusters":      clusters,
        "y_labels":      y_labels,
        "mean_N":        mean_N,
        "mean_A":        mean_A,
        "std_N":         std_N,
        "std_A":         std_A,
        "X_N_raw":       X_N_raw,
        "X_A_raw":       X_A_raw,
    }
    print(f"  ARI={ari:.3f}  sil={sil:.3f}  overlap={overlap:.3f}  "
          f"frac_sig={frac_sig:.2f}  wb={wb_ratio:.2f}")

# Summary dataframe
df_sum = pd.DataFrame([
    {k: v for k, v in d.items()
     if k not in ('X_2d','X_pca','clusters','y_labels',
                  'mean_N','mean_A','std_N','std_A','X_N_raw','X_A_raw')}
    for d in patient_data.values()
], index=patient_data.keys())
df_sum.index.name = "patient"
df_sum.to_csv(os.path.join(fig_dir, "low_perf_metrics.csv"))
print("\n", df_sum[["group","ARI","silhouette","overlap","frac_sig_tp","wb_ratio"]].to_string())

# ==========================================================
# FIGURE 1 — t-SNE grid: Good (top row) vs Poor (bottom rows)
# Each panel: N=blue circles, A=red triangles, noise=gray
# Convex hulls drawn around each class
# ==========================================================

from matplotlib.patches import Ellipse
from scipy.spatial import ConvexHull

def draw_convex_hull(ax, points, color, alpha=0.12):
    if len(points) < 4:
        return
    try:
        hull = ConvexHull(points)
        verts = np.append(hull.vertices, hull.vertices[0])
        ax.fill(points[verts,0], points[verts,1], color=color, alpha=alpha)
        ax.plot(points[verts,0], points[verts,1], color=color,
                linewidth=0.8, alpha=0.5)
    except Exception:
        pass

n_good = len(GOOD_PATIENTS)
n_poor = len(POOR_PATIENTS)
n_cols = 4
n_rows = int(np.ceil(len(ALL_PATIENTS) / n_cols))

fig = plt.figure(figsize=(n_cols * 3.6, n_rows * 3.4))
gs  = gridspec.GridSpec(n_rows, n_cols, figure=fig, hspace=0.5, wspace=0.3)

ordered = GOOD_PATIENTS + POOR_PATIENTS

for pi, patient in enumerate(ordered):
    if patient not in patient_data:
        continue
    d   = patient_data[patient]
    row = pi // n_cols
    col = pi %  n_cols
    ax  = fig.add_subplot(gs[row, col])

    X2  = d["X_2d"]
    cl  = d["clusters"]
    yl  = d["y_labels"]

    mask_noise = cl == -1
    mask_N     = (~mask_noise) & (yl == 'N')
    mask_A     = (~mask_noise) & (yl == 'A')

    # Convex hulls
    if mask_N.sum() >= 4:
        draw_convex_hull(ax, X2[mask_N], CLR_N, alpha=0.15)
    if mask_A.sum() >= 4:
        draw_convex_hull(ax, X2[mask_A], CLR_A, alpha=0.15)

    if mask_noise.any():
        ax.scatter(X2[mask_noise,0], X2[mask_noise,1],
                   c=CLR_NOISE, s=8, alpha=0.35, zorder=1)
    ax.scatter(X2[mask_N,0], X2[mask_N,1],
               c=CLR_N, s=14, alpha=0.65, zorder=2)
    ax.scatter(X2[mask_A,0], X2[mask_A,1],
               c=CLR_A, s=14, alpha=0.65, marker='^', zorder=3)

    grp_color = "#2a9d8f" if d["group"] == "Good" else "#e76f51"
    ax.set_title(f"{patient}  [{d['group']}]",
                 fontsize=9, fontweight='bold', color=grp_color)
    ax.set_xlabel(
        f"ARI={d['ARI']:.2f}  sil={d['silhouette']:.2f}",
        fontsize=7.5
    )
    ax.set_xticks([]); ax.set_yticks([])
    for sp in ax.spines.values():
        sp.set_edgecolor(grp_color); sp.set_linewidth(2)

# Add row labels
good_row_mid = (0.5 * len(GOOD_PATIENTS) / n_cols) / n_rows
poor_row_mid = 1 - (0.5 * (1 + len(POOR_PATIENTS) / n_cols)) / n_rows
fig.text(0.01, 1 - good_row_mid, "Good", va='center', rotation=90,
         fontsize=10, fontweight='bold', color='#2a9d8f')
fig.text(0.01, poor_row_mid, "Poor", va='center', rotation=90,
         fontsize=10, fontweight='bold', color='#e76f51')

handles = [
    plt.Line2D([0],[0], marker='o', color='w', markerfacecolor=CLR_N,    markersize=8, label='Normal'),
    plt.Line2D([0],[0], marker='^', color='w', markerfacecolor=CLR_A,    markersize=8, label='Arrhythmic'),
    plt.Line2D([0],[0], marker='o', color='w', markerfacecolor=CLR_NOISE, markersize=8, label='Noise'),
]
fig.legend(handles=handles, loc='lower center', ncol=3,
           fontsize=9, bbox_to_anchor=(0.5, -0.02))
fig.suptitle("t-SNE: N vs A Separation — Good vs Poor Patients\n"
             "(shaded region = convex hull per class)", fontsize=12)

plt.savefig(os.path.join(fig_dir, "Fig1_tsne_good_vs_poor.pdf"), dpi=300, bbox_inches='tight')
plt.savefig(os.path.join(fig_dir, "Fig1_tsne_good_vs_poor.png"), dpi=300, bbox_inches='tight')
plt.show()

# ==========================================================
# FIGURE 2 — Mean waveform ± std: Good vs Poor patients
# 4 panels good, 8 panels poor
# Shows morphological similarity/difference between N and A
# ==========================================================

ordered_wf = GOOD_PATIENTS + POOR_PATIENTS
n_wf = len(ordered_wf)
n_cols_wf = 4
n_rows_wf = int(np.ceil(n_wf / n_cols_wf))
t_axis = np.arange(beat_length)

fig, axes = plt.subplots(n_rows_wf, n_cols_wf,
                         figsize=(n_cols_wf * 3.8, n_rows_wf * 3.0),
                         sharex=True)
axes = axes.flatten()

for pi, patient in enumerate(ordered_wf):
    if patient not in patient_data:
        continue
    d  = patient_data[patient]
    ax = axes[pi]

    ax.fill_between(t_axis, d["mean_N"]-d["std_N"], d["mean_N"]+d["std_N"],
                    alpha=0.18, color=CLR_N)
    ax.fill_between(t_axis, d["mean_A"]-d["std_A"], d["mean_A"]+d["std_A"],
                    alpha=0.18, color=CLR_A)
    ax.plot(t_axis, d["mean_N"], color=CLR_N, linewidth=1.5, label='Normal')
    ax.plot(t_axis, d["mean_A"], color=CLR_A, linewidth=1.5,
            linestyle='--', label='Arrhythmic')

    grp_color = "#2a9d8f" if d["group"] == "Good" else "#e76f51"
    ax.set_title(f"{patient} [{d['group']}]",
                 fontsize=9, fontweight='bold', color=grp_color)
    ax.set_xlabel(
        f"ARI={d['ARI']:.2f}  sig_tp={d['frac_sig_tp']:.0%}",
        fontsize=7.5
    )
    ax.set_ylabel("Amplitude", fontsize=7)
    ax.tick_params(labelsize=7)

    for sp in ax.spines.values():
        sp.set_edgecolor(grp_color); sp.set_linewidth(1.5)

for ax in axes[n_wf:]:
    ax.set_visible(False)

axes[0].legend(fontsize=8, loc='upper right')
fig.suptitle("Mean ± Std Waveform: Normal vs Arrhythmic\n"
             "Good patients show clear morphological difference;\n"
             "Poor patients show similar waveforms across classes",
             fontsize=11)
plt.tight_layout()
plt.savefig(os.path.join(fig_dir, "Fig2_mean_waveform_good_vs_poor.pdf"), dpi=300, bbox_inches='tight')
plt.savefig(os.path.join(fig_dir, "Fig2_mean_waveform_good_vs_poor.png"), dpi=300, bbox_inches='tight')
plt.show()

# ==========================================================
# FIGURE 3 — Quantitative metrics comparison: Good vs Poor
# 4 subplots: ARI, Silhouette, Overlap, Frac sig timepoints
# ==========================================================

metrics = [
    ("ARI",          "ARI\n(0=random, 1=perfect)",      True),
    ("silhouette",   "Silhouette Score\n(1=well separated, -1=wrong cluster)", True),
    ("overlap",      "Class Overlap\n(1=total overlap, 0=separated)",  False),
    ("frac_sig_tp",  "Fraction Significant\nTimepoints (Mann-Whitney)", True),
]

fig, axes = plt.subplots(1, 4, figsize=(14, 5))

for ax, (col, ylabel, higher_better) in zip(axes, metrics):
    good_vals = [patient_data[p][col] for p in GOOD_PATIENTS if p in patient_data]
    poor_vals = [patient_data[p][col] for p in POOR_PATIENTS if p in patient_data]

    # Individual points
    jitter_g = np.random.default_rng(0).uniform(-0.08, 0.08, len(good_vals))
    jitter_p = np.random.default_rng(1).uniform(-0.08, 0.08, len(poor_vals))
    ax.scatter(np.ones(len(good_vals))  + jitter_g, good_vals,
               color="#2a9d8f", s=50, alpha=0.8, zorder=3)
    ax.scatter(np.full(len(poor_vals),2) + jitter_p, poor_vals,
               color="#e76f51", s=50, alpha=0.8, zorder=3)

    # Mean ± std bars
    for xi, vals, col_bar in [(1, good_vals, "#2a9d8f"), (2, poor_vals, "#e76f51")]:
        m, s = np.mean(vals), np.std(vals)
        ax.errorbar(xi, m, yerr=s, fmt='_', color=col_bar,
                    markersize=18, markeredgewidth=2.5, capsize=6, linewidth=2, zorder=4)

    # Annotate each point with patient ID
    for xi, vals, pats, jit in [
        (1, good_vals, GOOD_PATIENTS, jitter_g),
        (2, poor_vals, POOR_PATIENTS, jitter_p)
    ]:
        for v, pat, j in zip(vals, pats, jit):
            ax.annotate(pat, (xi+j, v), fontsize=6,
                        xytext=(4, 2), textcoords='offset points', color='dimgray')

    # Mann-Whitney test
    if len(good_vals) > 1 and len(poor_vals) > 1:
        _, p_val = mannwhitneyu(good_vals, poor_vals, alternative='two-sided')
        sig = "***" if p_val < 0.001 else "**" if p_val < 0.01 else "*" if p_val < 0.05 else "ns"
        y_max = max(max(good_vals), max(poor_vals))
        ax.annotate(f"p={p_val:.3f} {sig}",
                    xy=(1.5, y_max * 1.05), ha='center', fontsize=8,
                    color='black')
        ax.plot([1, 2], [y_max * 1.03, y_max * 1.03], color='black', linewidth=0.8)

    ax.set_xticks([1, 2])
    ax.set_xticklabels(["Good\n(n=11)", "Poor\n(n=8)"], fontsize=10)
    ax.set_ylabel(ylabel, fontsize=9)
    ax.set_xlim(0.5, 2.5)
    ax.grid(True, axis='y', alpha=0.3)
    arrow = "↑ better" if higher_better else "↓ better"
    ax.set_title(arrow, fontsize=8, color='gray')

fig.suptitle("Quantitative Evidence: Good vs Poor N/A Separability", fontsize=12)
plt.tight_layout()
plt.savefig(os.path.join(fig_dir, "Fig3_metrics_good_vs_poor.pdf"), dpi=300, bbox_inches='tight')
plt.savefig(os.path.join(fig_dir, "Fig3_metrics_good_vs_poor.png"), dpi=300, bbox_inches='tight')
plt.show()

# ==========================================================
# FIGURE 4 — PCA scatter (PC1 vs PC2) per patient
# More interpretable than t-SNE for showing linear separability
# ==========================================================

fig, axes = plt.subplots(n_rows_wf, n_cols_wf,
                         figsize=(n_cols_wf * 3.6, n_rows_wf * 3.2))
axes = axes.flatten()

for pi, patient in enumerate(ordered_wf):
    if patient not in patient_data:
        continue
    d   = patient_data[patient]
    ax  = axes[pi]
    Xpc = d["X_pca"]
    yl  = d["y_labels"]

    ax.scatter(Xpc[yl=='N', 0], Xpc[yl=='N', 1],
               c=CLR_N, s=12, alpha=0.55, label='Normal')
    ax.scatter(Xpc[yl=='A', 0], Xpc[yl=='A', 1],
               c=CLR_A, s=12, alpha=0.55, marker='^', label='Arrhythmic')

    grp_color = "#2a9d8f" if d["group"] == "Good" else "#e76f51"
    ax.set_title(f"{patient} [{d['group']}]",
                 fontsize=9, fontweight='bold', color=grp_color)
    ax.set_xlabel(f"sil={d['silhouette']:.2f}  overlap={d['overlap']:.2f}", fontsize=7.5)
    ax.set_ylabel("PC2", fontsize=7); ax.set_xlabel("PC1", fontsize=7)
    ax.tick_params(labelsize=6)
    for sp in ax.spines.values():
        sp.set_edgecolor(grp_color); sp.set_linewidth(1.5)

for ax in axes[n_wf:]:
    ax.set_visible(False)

axes[0].legend(fontsize=7, loc='upper right')
fig.suptitle("PCA (PC1 vs PC2): N vs A Overlap per Patient\n"
             "Good patients show clear separation in PC space;\n"
             "Poor patients show heavily overlapping clouds",
             fontsize=11)
plt.tight_layout()
plt.savefig(os.path.join(fig_dir, "Fig4_pca_scatter_good_vs_poor.pdf"), dpi=300, bbox_inches='tight')
plt.savefig(os.path.join(fig_dir, "Fig4_pca_scatter_good_vs_poor.png"), dpi=300, bbox_inches='tight')
plt.show()

# ==========================================================
# FIGURE 5 — Waveform difference map (heatmap)
# For each patient: heatmap of |mean_N - mean_A| over time
# Good patients → large difference. Poor → near zero.
# ==========================================================

fig, axes = plt.subplots(2, 1, figsize=(14, 6),
                         gridspec_kw={'height_ratios': [1, 1]})

for ax, group, patients, grp_color in [
    (axes[0], "Good", GOOD_PATIENTS, "#2a9d8f"),
    (axes[1], "Poor", POOR_PATIENTS, "#e76f51"),
]:
    pat_list = [p for p in patients if p in patient_data]
    diff_mat  = np.array([
        np.abs(patient_data[p]["mean_N"] - patient_data[p]["mean_A"])
        for p in pat_list
    ])

    im = ax.imshow(diff_mat, aspect='auto', cmap='YlOrRd',
                   vmin=0, vmax=diff_mat.max())
    ax.set_yticks(range(len(pat_list)))
    ax.set_yticklabels(pat_list, fontsize=9)
    ax.set_xlabel("Sample index (0→180)", fontsize=9)
    ax.set_title(f"{group} patients — |mean_N − mean_A| over time",
                 fontsize=10, color=grp_color, fontweight='bold')
    plt.colorbar(im, ax=ax, label="|diff|", fraction=0.02)

fig.suptitle("Morphological Difference Between N and A Waveforms\n"
             "Good patients show large differences; Poor patients show near-zero difference",
             fontsize=11)
plt.tight_layout()
plt.savefig(os.path.join(fig_dir, "Fig5_waveform_diff_heatmap.pdf"), dpi=300, bbox_inches='tight')
plt.savefig(os.path.join(fig_dir, "Fig5_waveform_diff_heatmap.png"), dpi=300, bbox_inches='tight')
plt.show()

# ==========================================================
# PRINT PAPER-READY NUMBERS
# ==========================================================

print("\n" + "="*60)
print("PAPER-READY NUMBERS")
print("="*60)

for grp, pats in [("Good", GOOD_PATIENTS), ("Poor", POOR_PATIENTS)]:
    vals = {m: [patient_data[p][m] for p in pats if p in patient_data]
            for m in ["ARI","silhouette","overlap","frac_sig_tp","wb_ratio"]}
    print(f"\n{grp} patients ({len(pats)}): {pats}")
    for m, v in vals.items():
        print(f"  {m:15s}: {np.mean(v):.3f} ± {np.std(v):.3f}")

# Statistical tests
print("\n--- Mann-Whitney U (Good vs Poor) ---")
for m in ["ARI", "silhouette", "overlap", "frac_sig_tp"]:
    g = [patient_data[p][m] for p in GOOD_PATIENTS if p in patient_data]
    po = [patient_data[p][m] for p in POOR_PATIENTS if p in patient_data]
    _, p_val = mannwhitneyu(g, po, alternative='two-sided')
    print(f"  {m:15s}: p = {p_val:.4f}")
```
